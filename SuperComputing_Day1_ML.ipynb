```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#command to load dataset via seaborn

titanic = sns.load_dataset('titanic')
```

```python
titanic.head()
```

```python
print("Shape of Dataset:")
print(titanic.shape)

print("\nColumn Names:")
print(titanic.columns)

print("\nDataset Info:")
print(titanic.info())
```

```python
print("Numerical Summary:")
print(titanic.describe())
```

print(titanic.describe(include='object'))


What does it do?

describe() normally gives statistics for numerical columns only.

When you write:

include='object'

it gives a summary of categorical/string columns such as:

sex
embarked
class
who
embark_town
alive

```python
print("\nCategorical Summary:")
print(titanic.describe(include='object'))
```

```python
plt.figure(figsize=(5,4))
sns.countplot(x='survived', data=titanic)
plt.title('Survived Count')
plt.show()


print("Survival Counts:")
print(titanic['survived'].value_counts())

```

```python
numerical_cols=['pclass','age','sibsp','parch','fare']

for cols in numerical_cols:
  plt.figure(figsize=(5,4))
  sns.histplot(titanic[cols],kde=True,bins=20)
  plt.title(f'Distribution of {cols}')
  plt.show()
```

```python
categorical=['sex', 'embarked','who','embark_town','alive']
for cols in categorical:
  plt.figure(figsize=(5,4))
  sns.countplot(x=cols,data=titanic)
  plt.title(f'Distribution of {cols}')
  plt.show()

```

```python
# Survival vs Gender
plt.figure(figsize=(6,4))
sns.countplot(x='sex', hue='survived', data=titanic)
plt.title("Survival by Gender")
plt.show()

# Survival vs Passenger Class
plt.figure(figsize=(6,4))
sns.countplot(x='pclass', hue='survived', data=titanic)
plt.title("Survival by Passenger Class")
plt.show()

# Survival vs Embarked
plt.figure(figsize=(6,4))
sns.countplot(x='embarked', hue='survived', data=titanic)
plt.title("Survival by Embarked Port")
plt.show()
```

```python
for col in numerical_cols:

    plt.figure(figsize=(8,5))

    sns.boxplot(

        x='survived',

        y=col,

        data=titanic

    )

    plt.title(f"{col} by Species")

    plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(

    titanic[numerical_cols].corr(),

    annot=True,

    cmap='coolwarm',

    fmt='.2f'

)

plt.title("Correlation Matrix")

plt.show()
```

```python
sns.pairplot(

    titanic,

    hue='survived',

    diag_kind='hist'

)

plt.show()
```
